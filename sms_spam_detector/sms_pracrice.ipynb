```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer

from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
df=pd.read_csv("spam.csv",encoding="latin-1")
df.head()
```

```python
df.columns
```

```python
df.shape
```

```python
df=df[["v1","v2"]]
df.columns=["label","message"]
df.head()
```

```python
df=df.dropna()
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["label"]=df["label"].map({"ham":0,"spam":1})
```

```python
df["label"].value_counts()
```

```python
x=df["message"]
y=df["label"]
```

```python
x
```

```python
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(
    x,y,test_size=0.2,random_state=42,stratify=y)
```

```python
vactorizer=TfidfVectorizer(lowercase=True,stop_words="english")
x_train=vactorizer.fit_transform(x_train)
x_test=vactorizer.transform(x_test)
                           
```

```python
model=MultinomialNB()
model.fit(x_train,y_train)
```

```python
y_pred=model.predict(x_test)

```

```python
accuracy=accuracy_score(y_test,y_pred)
print(accuracy)
```

```python
cm=confusion_matrix(y_test,y_pred)
print(cm)
```

```python
cr=classification_report(y_test,y_pred,target_names=["ham","spam"])
print(cr)
```

```python
always_ham=np.zeros(len(y_test))
baseline_accuracy=accuracy_score(
    y_test,
    always_ham)

print(baseline_accuracy)
print(accuracy)
```
