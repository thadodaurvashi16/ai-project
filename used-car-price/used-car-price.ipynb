```python
df["fuel_type"] = df["fuel_type"].fillna("Unknown")
df["accident"] = df["accident"].fillna("Unknown")
df["clean_title"] = df["clean_title"].fillna("Unknown")
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df = df.drop_duplicates()
```

```python
print("Duplicate rows:", df.duplicated().sum())
```

```python
df["car_age"] = 2026 - df["model_year"]
```

```python
print(df[["model_year", "car_age"]].head())
```

```python
# Convert milage to numeric
df["milage"] = pd.to_numeric(df["milage"], errors="coerce")

# Create mileage per year
df["mileage_per_year"] = (
    df["milage"] / df["car_age"].replace(0, 1)
)
```

```python
df["engine_liters"] = df["engine"].str.extract(r"(\d+\.\d+)L")[0]
```

```python
df["engine_liters"] = pd.to_numeric(
    df["engine_liters"],
    errors="coerce"
)
```

```python
print(df[["engine", "engine_liters"]].head())
```

```python
categorical_columns = df.select_dtypes(
    include="object"
).columns

print(categorical_columns)
```

```python
numeric_columns = df.select_dtypes(
    include=np.number
).columns

print(numeric_columns)
```
