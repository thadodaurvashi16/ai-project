```python
df = df.drop_duplicates()
```

```python
print("\nShape after removing duplicates:")
print(df.shape)
```

```python
df["model_year"] = pd.to_numeric(
    df["model_year"],
    errors="coerce"
)
```

```python
df["milage"] = (
    df["milage"]
    .astype(str)
    .str.replace(r"[^\d.]", "", regex=True)
)
```

```python
df["milage"] = pd.to_numeric(
    df["milage"],
    errors="coerce"
)
```

```python
df["price"] = (
    df["price"]
    .astype(str)
    .str.replace(r"[^\d.]", "", regex=True)
)
```

```python

df["price"] = pd.to_numeric(
    df["price"],
    errors="coerce"
)
```

```python
df = df.dropna(subset=["price"])
```

```python
df["model_year"] = df["model_year"].fillna(
    df["model_year"].median()
)
```

```python
df["milage"] = df["milage"].fillna(
    df["milage"].median()
)
```

```python
categorical_columns = [
    "brand",
    "model",
    "fuel_type",
    "engine",
    "transmission",
    "ext_col",
    "int_col",
    "accident",
    "clean_title"
]

for col in categorical_columns:
    df[col] = df[col].fillna(
        df[col].mode()[0]
    )
```

```python
CURRENT_YEAR = 2026

df["car_age"] = (
    CURRENT_YEAR - df["model_year"]
)
```

```python
df["engine_liters"] = pd.to_numeric(
    df["engine"]
    .astype(str)
    .str.extract(r"(\d+\.\d+)")[0],
    errors="coerce"
)
```

```python
df["engine_liters"]
```

```python
df["mileage_per_year"] = (
    df["milage"] / (df["car_age"] + 1)
)
```

```python
df["mileage_per_year"] 
```

```python
df["engine_liters"] = df["engine_liters"].fillna(
    df["engine_liters"].median()
)
```

```python

df = df[df["car_age"] >= 0].copy()
```
